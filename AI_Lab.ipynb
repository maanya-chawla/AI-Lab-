```python
#TicTacToe
import random

board = [" " for _ in range(9)]

def display_board():
    print()
    print(board[0], "|", board[1], "|", board[2])
    print("--+---+--")
    print(board[3], "|", board[4], "|", board[5])
    print("--+---+--")
    print(board[6], "|", board[7], "|", board[8])
    print()

def check_winner(player):
    combinations = [
        (0, 1, 2),
        (3, 4, 5),
        (6, 7, 8),
        (0, 3, 6),
        (1, 4, 7),
        (2, 5, 8),
        (0, 4, 8),
        (2, 4, 6)
    ]

    for a, b, c in combinations:
        if board[a] == player and board[b] == player and board[c] == player:
            return True

    return False

def ai_move():
    empty_positions = []

    for i in range(9):
        if board[i] == " ":
            empty_positions.append(i)

    if not empty_positions:
        return

    # 1. Check if AI can win in the next move
    for i in empty_positions:
        board[i] = "O"
        if check_winner("O"):
            print("AI chose position:", i + 1)
            return
        board[i] = " " # Undo the move

    # 2. Check if player can win in the next move and block them
    for i in empty_positions:
        board[i] = "X"
        if check_winner("X"):
            board[i] = "O" # Block the player
            print("AI chose position:", i + 1)
            return
        board[i] = " " # Undo the move

    # 3. Try to take the center
    if 4 in empty_positions:
        board[4] = "O"
        print("AI chose position: 5")
        return

    # 4. Try to take a corner
    corners = [0, 2, 6, 8]
    available_corners = [c for c in corners if c in empty_positions]
    if available_corners:
        position = random.choice(available_corners)
        board[position] = "O"
        print("AI chose position:", position + 1)
        return

    # 5. Take any remaining empty spot (sides)
    if empty_positions:
        position = random.choice(empty_positions)
        board[position] = "O"
        print("AI chose position:", position + 1)

def game():
    print("TIC-TAC-TOE")
    print("You = X")
    print("AI  = O")

    for turn in range(9):

        display_board()

        # Human's turn
        if turn % 2 == 0:
            while True:
                try:
                    position = int(input("Enter your position (1-9): "))
                    if position < 1 or position > 9:
                        print("Invalid position!")
                        continue
                    if board[position - 1] != " ":
                        print("Position already occupied!")
                        continue
                    break
                except ValueError:
                    print("Invalid input. Please enter a number between 1 and 9.")

            board[position - 1] = "X"

            if check_winner("X"):
                display_board()
                print("You win!")
                return

        # AI's turn
        else:
            ai_move()

            if check_winner("O"):
                display_board()
                print("AI wins!")
                return

    display_board()
    print("It's a draw!")

game()
```

```python
import random

class TwoRoomEnvironment:
    def __init__(self):
        self.rooms = {'A': random.choice(['Clean', 'Dirty']),
                      'B': random.choice(['Clean', 'Dirty'])}
        print("Initial Environment State:", self.rooms)

    def is_dirty(self, room_location):
        return self.rooms[room_location] == 'Dirty'

    def clean_room(self, room_location):
        if self.is_dirty(room_location):
            self.rooms[room_location] = 'Clean'
            return True
        return False

    def display_environment(self, agent_location=None):
        # Simplified display
        agent_in_a = '<-- Agent' if agent_location == 'A' else ''
        agent_in_b = '<-- Agent' if agent_location == 'B' else ''
        print(f"Rooms: A: {self.rooms['A']} {agent_in_a}, B: {self.rooms['B']} {agent_in_b}")

class SimpleVacuumAgent:
    def __init__(self, env):
        self.env = env
        self.location = random.choice(['A', 'B'])
        self.score = 0

    def perceive(self):
        return self.location, self.env.is_dirty(self.location)

    def act(self):
        current_location, is_dirty = self.perceive()

        if is_dirty:
            # print(f"Agent in Room {current_location} perceives it's Dirty. Cleaning...")
            if self.env.clean_room(current_location):
                self.score += 1
            return "Suck"
        else:
            # print(f"Agent in Room {current_location} perceives it's Clean. Moving...")
            self.location = 'B' if current_location == 'A' else 'A'
            return "Move"

    def run_agent(self, steps=10):
        print(f"Agent starts in Room {self.location}")
        for step in range(steps):
            # self.env.display_environment(self.location) # Display only at end or for key events
            action = self.act()
            print(f"Step {step + 1}: Agent in {self.location}, Action: {action}, Score: {self.score}, Env: {self.env.rooms}")
            # time.sleep(0.1) # Removed for faster execution

        print("\n--- Simulation Complete ---")
        self.env.display_environment(self.location)
        print(f"Total dirt cleaned: {self.score} spots")


# --- Run the Simulation ---
env = TwoRoomEnvironment()
agent = SimpleVacuumAgent(env)
agent.run_agent(steps=10)
```

```python
class PuzzleState:
    def __init__(self, board, parent=None, move=None):
        self.board = board # board is a tuple of tuples (immutable and hashable)
        self.parent = parent
        self.move = move
        self.empty_pos = self._find_empty_position()

    def _find_empty_position(self):
        for r in range(3):
            for c in range(3):
                if self.board[r][c] == 0:
                    return r, c
        return None

    def __eq__(self, other):
        return self.board == other.board

    def __hash__(self):
        return hash(self.board)

    def __str__(self):
        s = ""
        for row in self.board:
            s += " ".join(map(str, row)).replace('0', ' ')
            s += "\n"
        return s

    def get_successors(self):
        successors = []
        er, ec = self.empty_pos
        moves = [(0, 1, 'Right'), (0, -1, 'Left'), (1, 0, 'Down'), (-1, 0, 'Up')]

        for dr, dc, move_name in moves:
            nr, nc = er + dr, ec + dc
            if 0 <= nr < 3 and 0 <= nc < 3:
                new_board_list = [list(row) for row in self.board]
                new_board_list[er][ec], new_board_list[nr][nc] = new_board_list[nr][nc], new_board_list[er][ec]
                new_board_tuple = tuple(tuple(row) for row in new_board_list)
                successors.append(PuzzleState(new_board_tuple, self, move_name))
        return successors
```

```python
def dfs_solver(initial_state, goal_state):
    stack = [initial_state]
    visited = {initial_state.board}

    while stack:
        current_state = stack.pop()

        if current_state == goal_state:
            path = []
            temp_state = current_state
            while temp_state.parent:
                path.append(temp_state.move)
                temp_state = temp_state.parent
            path.reverse()
            return path, current_state

        for successor in current_state.get_successors():
            if successor.board not in visited:
                visited.add(successor.board)
                stack.append(successor)

    return None, None
```

```python
# --- Demonstration ---

initial_board = (
    (2, 8, 3),
    (1, 6, 4),
    (7, 0, 5) # 0 represents the empty tile
)

goal_board = (
    (1, 2, 3),
    (8, 0, 4),
    (7, 6, 5) # 0 represents the empty tile
)

initial_state = PuzzleState(initial_board)
goal_state = PuzzleState(goal_board)

print("Initial State:")
print(initial_state)

print("Goal State:")
print(goal_state)

print("Searching for solution...")
path, final_state_reached = dfs_solver(initial_state, goal_state)

if path:
    print(f"Solution Found in {len(path)} moves:")
    print("Moves:", path)
    print("\nFinal Board State:")
    print(final_state_reached)
else:
    print("No solution found.")
```

```python
def dls_solver(initial_state, goal_state, limit):
    stack = [(initial_state, 0)] # (state, current_depth)
    # Visited set helps avoid redundant work within a single DLS call
    # and cycles, important for state-space search.
    visited = {initial_state.board}

    while stack:
        current_state, current_depth = stack.pop()

        if current_state == goal_state:
            path = []
            temp_state = current_state
            while temp_state.parent:
                path.append(temp_state.move)
                temp_state = temp_state.parent
            path.reverse()
            return path, current_state

        if current_depth < limit:
            # Get successors and add them to the stack
            # Reverse to explore in a consistent order if desired, or just iterate
            for successor in reversed(current_state.get_successors()):
                if successor.board not in visited:
                    visited.add(successor.board)
                    stack.append((successor, current_depth + 1))
    return None, None

def ids_solver(initial_state, goal_state, max_depth=20):
    print(f"Starting IDS search up to max depth {max_depth}...")
    for depth_limit in range(max_depth + 1):
        print(f"  Trying depth limit: {depth_limit}")
        # Each DLS call operates independently with its own visited set
        path, final_state = dls_solver(initial_state, goal_state, depth_limit)
        if path is not None:
            return path, final_state
    return None, None

# --- Demonstration with IDS ---
# Re-using initial_board and goal_board defined in cell d3e494a7

# Ensure initial_board and goal_board are defined if this cell is run independently
# If running the entire notebook, they will be from the previous cell.
# initial_board = (
#     (2, 8, 3),
#     (1, 6, 4),
#     (7, 0, 5) # 0 represents the empty tile
# )

# goal_board = (
#     (1, 2, 3),
#     (8, 0, 4),
#     (7, 6, 5) # 0 represents the empty tile
# )

# Assume PuzzleState class is available from a previous cell (fc96c1ef)
initial_state_ids = PuzzleState(initial_board)
goal_state_ids = PuzzleState(goal_board)

print("\n--- IDS Solver Demonstration ---")
print("Initial State for IDS:")
print(initial_state_ids)

print("Goal State for IDS:")
print(goal_state_ids)

path_ids, final_state_ids_reached = ids_solver(initial_state_ids, goal_state_ids)

if path_ids:
    print(f"\nIDS Solution Found in {len(path_ids)} moves (at depth {len(path_ids)}):")
    print("Moves:", path_ids)
    print("\nFinal Board State from IDS:")
    print(final_state_ids_reached)
else:
    print("\nNo solution found by IDS within the given max depth.")
```
